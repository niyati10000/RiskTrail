# Phase 1: Data Audit & Exploratory Data Analysis (EDA)
This notebook conducts the initial data audit for **RiskTrail** on credit sequence statement data.
- Customer count and sequence length distribution
- Target class balance
- Missingness per column and statement step
- Categorical cardinalities
- Temporal distribution and feasibility of time-aware checks

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import yaml

statements = pd.read_parquet('../data/interim/statements.parquet')
labels = pd.read_parquet('../data/interim/labels.parquet')
print(f'Statements shape: {statements.shape}')
print(f'Labels shape: {labels.shape}')
```

```python
# Customer history distribution
stmt_counts = statements.groupby('customer_ID').size()
print('Statements per customer summary:')
print(stmt_counts.describe())
print(f'Max statements: {stmt_counts.max()} (expected <= 13)')
```

```python
# Class balance
class_counts = labels['target'].value_counts(normalize=True) * 100
print('Target class distribution (%):')
print(class_counts)
```

```python
# Missingness per column
missing = statements.isna().mean() * 100
print('Features with missing values (%):')
print(missing[missing > 0].sort_values(ascending=False))
```

```python
# Load column configurations
with open('../configs/columns.yaml', 'r') as f:
    cols = yaml.safe_load(f)
print('Categorical features:', cols['categorical_features'])
print('Numeric features count:', len(cols['numeric_features']))
```
